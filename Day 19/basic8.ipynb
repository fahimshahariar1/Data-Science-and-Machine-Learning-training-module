```python
import numpy as np
a = np.array([10,20,30])
print(a)
```

```python
b = np.linspace(1, 10, num = 2)
print(b)
```

```python
c = np.arange(5)
print(c)
```

```python
x = np.zeros(5)
print(x)
#similar can be done with ones
```

```python
#2D array with Numpy
x = np.array([[2,4,6], [1,3,5]])
print(x)
```

```python
#ndim attribute
#to know the number of rows of an array
a = np.array([1,2,3])
x = np.array([[2,4,6], [1,3,5]])

print(a.ndim)
print(x.ndim)
```

```python
#shape attribute

x = np.array([[2,4,6], [1,3,5]])
print(x.shape) #2 rows 3 columns
```

```python
#size attribute

x = np.array([[2,4,6], [1,3,5]])
print(x.size)
```

```python
#reshape method
x = np.array([[2,4,6], [1,3,5]])
d = x.reshape(3,2)
print(d)
```

```python
#concatenate

a = np.array([1,2,3])
b = np.array([4,5,6])

c = np.concatenate([a,b])
print(c)
```

```python
#hstack to join more than one array horizontally or Row wise
a = np.array([1,2,3])
b = np.array([4,5,6])

c = np.hstack((a,b))
print(c)
```

```python
#vstack to join more than one array vertically or column wise

a = np.array([1,2,3])
b = np.array([4,5,6])

c = np.vstack((a,b))
print(c)
```
