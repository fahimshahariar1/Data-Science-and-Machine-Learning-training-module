```python
#pandas

import pandas as pd
```

```python
#creating dat frame using pandas

ages = [10, 20, 45, 25, 66, 97, 100]

pd.DataFrame(ages)
```

```python
df_ages = pd.DataFrame(ages)
df_ages.head()
#we can specify a number inside head(). If not it will show the first 5 values
```

```python
#adding name attribute

data = {
    "Name": ["Fahim", "Pranto", "Siyam", "Arif", "Sayem", "Nafi", "Tauhid"],
    "Age": ages
}

print(data)
```

```python
df_sample = pd.DataFrame(data)
#df_sample is a variable name it can be anything
```

```python
df_sample.head(3)
```

```python
df_sample.tail()
#for tail() it will show values from the end
```

```python
#grouping

df_sample.groupby('Name')
print(df_sample.groupby('Name').groups)
```
