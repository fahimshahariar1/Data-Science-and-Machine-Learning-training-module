```python
#matplotlib

import matplotlib
```

```python
import matplotlib.pyplot as plt
```

```python
# 2 methods to create plot

plt.plot()
```

```python
#basic plot
plt.plot([1,2,3,4,5])
```

```python
# adding axis

x = ([1,2,3,4,5])
y = ([5,2,3,1,4])

plt.plot(x,y);
```

```python
# second way

fig = plt.figure()
ax = fig.add_subplot()
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([1,2,3,4])
ax.plot(x,y)
plt.show()
```

```python
#workflow in matplotlib


x = [1,2,3,4,5]
y = [5,4,3,2,1]

fig, ax = plt.subplots(figsize =(10,10))
```

```python
# setting area of a figure

ax.plot(x,y)
```

```python
#customize plot
ax.set(title="Sample Data", xlabel = "x-axis", ylabel="y-axis")
fig.savefig("D:/Data Science/data.png")
plt.plot()

# using this you can directly set a location to save your data
```
