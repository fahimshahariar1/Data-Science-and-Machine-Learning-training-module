```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.arange(0, 4 * np.pi, 0.1)
y = np.cos(x)

# Create plot
plt.figure(figsize=(10, 8))
plt.plot(x, y, label="cos(x)")

# Add labels and title
plt.xlabel("x")
plt.ylabel("cos(x)")
plt.title("Plot of cos(x)")

# Add grid
plt.grid(True)

# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.arange(0, 4 * np.pi, 0.1)
y = np.cos(x)

# Create plot
plt.figure(figsize=(10, 6))  # Adjust figure size
plt.plot(x, y, color='darkblue', linestyle='dashed', linewidth=2, marker='o', markersize=5, label='cosine')

# Add labels and title
plt.xlabel('x', fontsize=12)
plt.ylabel('cos(x)', fontsize=12)
plt.title('Cosine Function Plot', fontsize=14, fontweight='bold')

# Add grid and legend
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()

# Set plot limits
plt.xlim(0, 4 * np.pi)
plt.ylim(-1.2, 1.2)

# Show the plot
plt.show()
```

```python
#bar plot
import matplotlib.pyplot as plt

products = ['A', 'B', 'C']
sales = [100, 150, 80]

plt.bar(products, sales)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sales')
plt.show()
```

```python
# horizontal bar plot
import matplotlib.pyplot as plt

# Data
countries = ['USA', 'Canada', 'UK', 'Australia']
population = [331, 38, 67, 25]

# Create the bar plot
plt.barh(countries, population)

# Add labels and title
plt.xlabel('Population (in millions)')
plt.ylabel('Country')
plt.title('Population of Selected Countries')

plt.show()

```

```python
#stacked bar plot
import matplotlib.pyplot as plt

# Data
years = ['2018', '2019', '2020']
revenue = [120, 150, 180]
cost = [80, 100, 120]

# Create the bar plot
plt.bar(years, revenue, label='Revenue')
plt.bar(years, cost, label='Cost')

# Add labels and title
plt.xlabel('Year')
plt.ylabel('Amount')
plt.title('Revenue and Cost by Year')
plt.legend()

plt.show()
```

```python
#grouped bar plot
import matplotlib.pyplot as plt

# Data
categories = ['Product A', 'Product B', 'Product C']
sales_2022 = [100, 120, 80]
sales_2023 = [120, 150, 90]

# Create the bar plot
x = np.arange(len(categories))
width = 0.35

plt.bar(x - width/2, sales_2022, width, label='2022')
plt.bar(x + width/2, sales_2023, width, label='2023')

plt.xticks(x, categories)
plt.xlabel('Product')
plt.ylabel('Sales')
plt.title('Product Sales by Year')
plt.legend()

plt.show()
```

```python
#histogram

import matplotlib.pyplot as plt
import numpy as np

# Sample data
data = np.random.randn(1000)

# Create the histogram
plt.hist(data, bins=30, density=True)

# Add labels and title
plt.xlabel('Value')
plt.ylabel('Density')
plt.title('Histogram of Random Data')

plt.show()
```

```python
#scatter plot
import matplotlib.pyplot as plt
import numpy as np

# Sample data
height = np.random.normal(170, 10, 100)
weight = np.random.normal(70, 15, 100)

# Create the scatter plot
plt.scatter(height, weight)

# Add labels and title
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.title('Height vs Weight')

plt.show()
```

```python
#pie chart

import matplotlib.pyplot as plt

# Data
expenses = {'Rent': 500, 'Food': 300, 'Transportation': 150, 'Entertainment': 100}

# Create the pie chart
plt.pie(expenses.values(), labels=expenses.keys(), autopct='%1.1f%%')

# Add title
plt.title('Monthly Expenses')

plt.show()
```
